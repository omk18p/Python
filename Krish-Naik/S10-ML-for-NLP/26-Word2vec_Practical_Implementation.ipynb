# Using Pretrained Word2Vec (Google News 300)

## Import Libraries

```python
import gensim
from gensim.models import Word2Vec, KeyedVectors
import gensim.downloader as api
```

- `gensim` → NLP library for Word2Vec and other embedding models.
- `Word2Vec` → Used to train your own Word2Vec model.
- `KeyedVectors` → Used to load pretrained word vectors.
- `gensim.downloader` → Downloads pretrained models.

---

## Load Pretrained Model

```python
wv = api.load('word2vec-google-news-300')
```

Downloads Google's pretrained Word2Vec model.

- Trained on **Google News dataset (~3 billion words)**.
- Each word is represented by a **300-dimensional vector**.

---

## Get Word Vector

```python
vec_king = wv['king']
```

Returns the 300-dimensional embedding for **"king"**.

---

## Display Vector

```python
vec_king
```

Displays the numerical embedding of **king**.

---

## Check Vector Dimension

```python
vec_king.shape
```

Output

```python
(300,)
```

Meaning every word is represented by **300 features**.

---

## Get Another Word Vector

```python
wv['cricket']
```

Returns the embedding vector for **cricket**.

---

## Find Similar Words

```python
wv.most_similar('cricket')
```

Returns words having embeddings closest to **cricket** using **Cosine Similarity**.

Example:

```text
cricketer
cricketing
batting
batsman
```

---

```python
wv.most_similar('happy')
```

Returns semantically similar words.

Example:

```text
joyful
glad
cheerful
delighted
```

---

## Compute Similarity

```python
wv.similarity("hockey", "sports")
```

Returns a similarity score between **0 and 1**.

- Close to **1** → Highly similar.
- Close to **0** → Less similar.

---

## Word Analogy

```python
vec = wv['king'] - wv['man'] + wv['woman']
```

Performs vector arithmetic.

Concept:

```text
King - Man + Woman ≈ Queen
```

---

## Find Closest Word

```python
wv.most_similar([vec])
```

Output (typically):

```text
Queen
```

This demonstrates that Word2Vec captures **semantic relationships** between words.

---

## Applications

- Find similar words.
- Word analogy.
- Semantic search.
- Recommendation systems.
- Chatbots.
- Question Answering.

---

## Key Points

- `word2vec-google-news-300` is a **pretrained Word2Vec model**.
- Each word is represented by a **300-dimensional vector**.
- Uses **Cosine Similarity** to find similar words.
- Supports **word analogy** (e.g., King − Man + Woman ≈ Queen).
- Pretrained models save training time and provide high-quality embeddings.

---

# Interview Questions

### Q1. What is `word2vec-google-news-300`?

**Answer:**  
It is Google's pretrained Word2Vec model trained on approximately **3 billion words** from the Google News dataset, where each word is represented by a **300-dimensional embedding**.

---

### Q2. What does `wv['king']` return?

**Answer:**  
It returns the **300-dimensional vector representation** (embedding) of the word **"king"**.

---

### Q3. What does `most_similar()` do?

**Answer:**  
It returns words whose embeddings are **most similar** to the given word using **Cosine Similarity**.

---

### Q4. What does `similarity(word1, word2)` return?

**Answer:**  
It returns the **Cosine Similarity score** between the embeddings of two words.

---

### Q5. Explain the analogy `King − Man + Woman ≈ Queen`.

**Answer:**  
Word2Vec learns semantic relationships between words. Vector arithmetic captures these relationships, so subtracting the concept of **man** from **king** and adding **woman** results in a vector close to **queen**.

---

### Q6. Why use a pretrained Word2Vec model instead of training from scratch?

**Answer:**  
Pretrained models are trained on **very large corpora**, provide high-quality embeddings, save computation time, and perform well for many NLP tasks.
