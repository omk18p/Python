````python
import pandas as pd
messages=pd.read_csv('smsspamcollection/SMSSpamCollection',
                    sep='\t',names=["label","message"])
````

````python
messages
````

````python
## Data Cleaning And Preprocessing
import re
import nltk
nltk.download('stopwords')
````

````python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
wordlemmatize=WordNetLemmatizer()
````

````python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[wordlemmatize.lemmatize(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
````

````python
corpus
````

## Create TF-IDF And NGrams

````python
from sklearn.feature_extraction.text import TfidfVectorizer
````

````python
tfidf=TfidfVectorizer(max_features=100)
X=tfidf.fit_transform(corpus).toarray()
````

````python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, 
    formatter=dict(float=lambda x: "%.3g" % x))
````

# TF-IDF Vectorization

This code converts the **cleaned corpus** into a **TF-IDF feature matrix**.

---

## Import TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

Imports the `TfidfVectorizer` class from Scikit-learn.

---

## Create TF-IDF Model

```python
tfidf = TfidfVectorizer(max_features=100)
```

### Parameter

- **max_features=100**
  - Keeps only the **100 most important words** based on TF-IDF scores.
  - Reduces feature size and computation.

---

## Fit and Transform

```python
X = tfidf.fit_transform(corpus).toarray()
```

- **fit()** → Learns the vocabulary and calculates IDF values.
- **transform()** → Converts text into TF-IDF vectors.
- **fit_transform()** → Performs both operations together.
- **toarray()** → Converts the sparse matrix into a NumPy array.

---

## Display Matrix

```python
X
```

Displays the TF-IDF feature matrix.

- **Rows** → Documents (messages).
- **Columns** → Words (features).
- **Values** → TF-IDF scores.

---

## NumPy Print Settings

```python
np.set_printoptions(...)
```

Formats the NumPy array for better readability without truncation.

---

## Pipeline

```text
Clean Corpus
      │
      ▼
TfidfVectorizer
      │
      ▼
Learn Vocabulary + IDF
      │
      ▼
Calculate TF-IDF Scores
      │
      ▼
TF-IDF Feature Matrix
      │
      ▼
Machine Learning Model
```

---

## Key Points

- Converts text into **TF-IDF vectors**.
- Gives **higher weight** to important (rare) words.
- Gives **lower weight** to common words.
- Produces a numerical feature matrix for Machine Learning.
- Generally performs **better than Bag of Words** for many NLP classification tasks.

### N-Grams

````python
tfidf=TfidfVectorizer(max_features=100,ngram_range=(2,2))
X=tfidf.fit_transform(corpus).toarray()
````
