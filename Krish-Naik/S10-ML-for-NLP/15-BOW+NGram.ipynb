````python
import pandas as pd
messages=pd.read_csv('smsspamcollection/SMSSpamCollection',
                    sep='\t',names=["label","message"])
````

````python
messages
````

````python
## Data Cleaning And Preprocessing
import re
import nltk
nltk.download('stopwords')
````

````python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
````

````python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
````

````python
corpus
````

# Text Preprocessing for Bag of Words

## Steps

1. **Create Corpus**
   - `corpus = []`
   - Stores cleaned text.

2. **Loop Through Messages**
   - Processes each message one by one.

3. **Remove Special Characters**
   ```python
   re.sub('[^a-zA-Z]', ' ', text)
   ```
   - Keeps only alphabets.

4. **Convert to Lowercase**
   ```python
   review.lower()
   ```
   - Treats "Hello" and "hello" as the same word.

5. **Tokenization**
   ```python
   review.split()
   ```
   - Splits sentence into words.

6. **Remove Stopwords + Apply Stemming**
   ```python
   ps.stem(word)
   ```
   - Removes common words and converts words to their root form.

7. **Join Words**
   ```python
   ' '.join(review)
   ```
   - Converts word list back into a sentence.

8. **Append to Corpus**
   ```python
   corpus.append(review)
   ```
   - Stores the cleaned sentence.

---

## Pipeline

```text
Raw Text
   ↓
Remove Special Characters
   ↓
Lowercase
   ↓
Tokenization
   ↓
Remove Stopwords
   ↓
Stemming
   ↓
Clean Corpus
```

> **Purpose:** Creates a cleaned corpus for feature extraction methods like **Bag of Words**, **TF-IDF**, and **Word2Vec**.

# Bag of Words (BoW) Model

## Create BoW

```python
from sklearn.feature_extraction.text import CountVectorizer
```

`CountVectorizer` converts text into a **Bag of Words (BoW)** feature matrix based on **word frequency**.

---

## Initialize CountVectorizer

```python
cv = CountVectorizer(max_features=100, binary=True)
```

### Parameters

- **max_features=100**
  - Keeps only the top **100 most frequent words**.
  - Reduces feature size and computation.

- **binary=True**
  - Uses **1** if a word is present.
  - Uses **0** if a word is absent.
  - Ignores word frequency.

---

## Fit and Transform

```python
X = cv.fit_transform(corpus).toarray()
```

- **fit()** → Learns the vocabulary.
- **transform()** → Converts text into numerical vectors.
- **fit_transform()** → Performs both steps together.
- **toarray()** → Converts the sparse matrix into a NumPy array.

---

## Display Matrix

```python
X
```

Shows the Bag of Words feature matrix.

Each:
- **Row** → One document.
- **Column** → One word from the vocabulary.
- **Value** → Word presence (Binary) or frequency.

---

## NumPy Display Settings

```python
np.set_printoptions(...)
```

Used only to display the entire matrix neatly without truncation.

---

## Pipeline

```text
Clean Corpus
      ↓
CountVectorizer
      ↓
Build Vocabulary
      ↓
Convert Text → Numerical Matrix
      ↓
Machine Learning Model
```

> **Remember:** With `binary=True`, BoW stores only **0 (Absent)** or **1 (Present)** instead of word counts.

# Bag of Words with N-grams

Instead of considering only **single words (unigrams)**, this model also considers **groups of consecutive words (N-grams)** to capture context.

---

## Create the Model

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=100,
    binary=True,
    ngram_range=(2,4)
)

X = cv.fit_transform(corpus).toarray()
```

---

## Parameters

### `max_features=100`

- Keeps only the **100 most frequent N-grams**.
- Reduces feature size and computation.

### `binary=True`

- `1` → N-gram is present.
- `0` → N-gram is absent.
- Ignores frequency.

### `ngram_range=(2,4)`

Generates:

- **Bigrams (2 words)**
- **Trigrams (3 words)**
- **Four-grams (4 words)**

Example:

Sentence:

```text
I love learning NLP
```

Generated N-grams:

**Bigrams**

```text
I love
love learning
learning NLP
```

**Trigrams**

```text
I love learning
love learning NLP
```

**Four-grams**

```text
I love learning NLP
```

---

## `fit_transform(corpus)`

- **fit()** → Learns the vocabulary (all N-grams).
- **transform()** → Converts text into numerical vectors.
- **fit_transform()** → Performs both operations together.

---

## `toarray()`

Converts the sparse matrix into a NumPy array.

---

## Pipeline

```text
Clean Corpus
      │
      ▼
Generate N-grams (2–4)
      │
      ▼
Build Vocabulary
      │
      ▼
Convert Text → Binary Feature Matrix
      │
      ▼
Machine Learning Model
```

---

## Difference

| Bag of Words | Bag of Words with N-grams |
|--------------|---------------------------|
| Uses single words | Uses groups of 2–4 consecutive words |
| Less context | Preserves partial context |
| Smaller vocabulary | Larger vocabulary |
| Faster | Slower |
| Lower memory | Higher memory |

> **Remember:** `ngram_range=(2,4)` creates **bigrams, trigrams, and four-grams**, helping the model capture phrases like **"not good"**, **"free entry now"**, or **"call customer service"** instead of treating each word independently.
