# One-Hot Representation in NLP

## Import

```python
from tensorflow.keras.preprocessing.text import one_hot
```

The `one_hot()` function is used to **convert each word in a sentence into a unique integer index**.

It is one of the simplest methods to represent text numerically before feeding it into a Deep Learning model.

---

# Step 1: Create Sentences

```python
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good'
]
```

These are the input sentences (corpus) that we want to convert into numerical form.

---

# Step 2: Define Vocabulary Size

```python
voc_size = 10000
```

`voc_size` represents the **maximum vocabulary size**.

- TensorFlow maps each word to an integer between **1** and **9999**.
- A larger vocabulary size reduces the chance of different words getting the same index (collision).

Example:

```text
Vocabulary Size = 10000

Possible Word IDs:

1
25
786
1543
9999
```

---

# Step 3: One-Hot Representation

```python
one_hot_repr = [one_hot(words, voc_size) for words in sent]
```

This line performs the following steps:

1. Takes one sentence at a time.
2. Splits the sentence into individual words.
3. Converts each word into a unique integer index using the specified vocabulary size.
4. Stores the encoded sentence as a list of integers.

---

# Example

Sentence:

```text
the glass of milk
```

Possible Output:

```text
[345, 7821, 126, 987]
```

Sentence:

```text
I am a good boy
```

Possible Output:

```text
[54, 621, 902, 1843, 321]
```

> **Note:** The exact numbers may change every time because TensorFlow uses a **hashing technique** to assign word indices.

---

# What Does `one_hot()` Actually Do?

It **does not create a traditional one-hot vector** like:

```text
Cat → [0,0,1,0,0]
```

Instead, TensorFlow's `one_hot()` function returns **integer word indices** using hashing.

Example:

```text
good      → 432
developer → 1987
glass     → 678
milk      → 999
```

These integer IDs are later used by an **Embedding Layer** to generate dense vector representations.

---

# Why Do We Need One-Hot Representation?

Deep Learning models cannot understand text directly.

For example:

```text
"I love AI"
```

A neural network cannot process words.

So we first convert them into numbers:

```text
"I love AI"

↓

[245, 983, 412]
```

Now the model can process the numerical data.

---

# Limitations

- Uses **hashing**, so two different words may sometimes get the same index (collision).
- Does not capture the **meaning** or **relationship** between words.
- Returns only integer IDs, not actual feature vectors.

Because of these limitations, **Embedding Layers** are commonly used after one-hot encoding.

---

# Workflow

```text
Sentence
      ↓
Split into Words
      ↓
One-Hot Encoding
      ↓
Integer IDs
      ↓
Embedding Layer
      ↓
Dense Word Vectors
      ↓
RNN / LSTM / GRU
```

---

# Key Points ⭐

- `one_hot()` converts words into **integer indices**.
- `voc_size` defines the maximum vocabulary size.
- TensorFlow uses **hashing** to generate word IDs.
- The output is a **list of integers**, not binary vectors.
- These integer IDs are used as input to the **Embedding Layer**.
- One-hot representation is an initial step in many NLP pipelines.

---

# Word Embedding Representation

## Import Required Libraries

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

### Explanation

- **Embedding** → Converts integer word IDs into dense vector representations (word embeddings).
- **pad_sequences** → Makes all input sequences the same length by adding padding.
- **Sequential** → Used to build a neural network layer by layer.
- **NumPy** → Used for numerical operations.

---

## Define Maximum Sentence Length

```python
sent_length = 8
```

Here, we specify that every sentence should have **8 positions**.

- If a sentence has fewer than 8 words, zeros will be added.
- If it has more than 8 words, extra words will be removed.

---

## Apply Padding

```python
embedded_docs = pad_sequences(
    one_hot_repr,
    padding='pre',
    maxlen=sent_length
)
```

### Explanation

- **one_hot_repr**
  - Contains the integer representation of each sentence obtained using `one_hot()`.

- **maxlen=8**
  - Ensures every sentence has a length of **8**.

- **padding='pre'**
  - Adds zeros at the **beginning** of shorter sentences.

Example:

Before padding

```text
[123, 456, 789]
```

After padding

```text
[0, 0, 0, 0, 0, 123, 456, 789]
```

If `padding='post'` were used, zeros would be added at the end.

---

## Display the Result

```python
print(embedded_docs)
```

Prints the padded sequences.

Example Output:

```text
[
 [0 0 0 0 123 456 789 222]
 [0 0 0 123 456 789 901 567]
 [0 0 123 852 789 145 321 654]
]
```

Now every sentence has the same length and is ready to be passed to the **Embedding Layer**.

---

## Workflow

```text
Sentence
    ↓
One-Hot Encoding
    ↓
Integer IDs
    ↓
Padding
    ↓
Equal Length Sequences
    ↓
Embedding Layer
```

---

## Key Points

- Neural networks require input sequences of **equal length**.
- `pad_sequences()` is used to make all sequences the same size.
- `maxlen` specifies the required sequence length.
- `padding='pre'` adds zeros at the beginning.
- The padded sequences become the input to the **Embedding Layer**.

# Feature Representation using Embedding Layer

## Define Embedding Dimension

```python
dim = 10
```

### Explanation

The embedding dimension defines the **size of the vector** that will represent each word.

Here,

```text
Embedding Dimension = 10
```

This means every word will be converted into a vector containing **10 numerical values**.

Example:

```text
good

↓

[0.12, -0.45, 0.87, ..., 0.23]
```

Instead of representing a word using a single number, the Embedding Layer learns a dense vector representation.

---

## Create the Model

```python
model = Sequential()
```

Creates an empty Sequential model.

Layers will be added one after another.

---

## Add the Embedding Layer

```python
model.add(
    Embedding(
        voc_size,
        dim,
        input_length=sent_length
    )
)
```

### Explanation of Parameters

### `voc_size`

Represents the vocabulary size.

Example:

```text
voc_size = 10000
```

The Embedding layer creates an **Embedding Matrix** of size:

```text
10000 × 10
```

That means every one of the 10,000 possible words gets its own 10-dimensional vector.

---

### `dim`

Specifies the embedding dimension.

```text
dim = 10
```

Each word is represented by a vector of length **10**.

---

### `input_length`

```text
input_length = sent_length
```

Specifies the length of every input sentence.

Since

```python
sent_length = 8
```

every input sentence contains **8 word positions**.

---

## Compile the Model

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

### Explanation

Since we are only testing the Embedding Layer, compilation is required before prediction.

- **Optimizer:** Adam
- **Loss Function:** Mean Squared Error (MSE)

At this stage, these values are not important because we are not training the model.

---

## View Model Summary

```python
model.summary()
```

Displays information about the model.

Example:

```text
Embedding Layer

Input Shape : (None, 8)

Output Shape : (None, 8, 10)

Parameters : 100000
```

### Understanding the Output Shape

```text
(None, 8, 10)
```

- **None** → Batch size (can be any number of sentences)
- **8** → Number of words in each sentence
- **10** → Embedding dimension

This means each sentence is converted into an **8 × 10 matrix**.

---

## Generate Word Embeddings

```python
model.predict(embedded_docs)
```

The Embedding Layer converts every integer word ID into a dense vector.

Example

Input

```text
[0, 0, 0, 12, 54, 98, 32, 65]
```

Output

```text
[
 [0.11, 0.45, ...],
 [0.23, 0.71, ...],
 ...
]
```

Each integer becomes a vector of length **10**.

Example

## Check Model Input Shape

```python

```

Output

```text

```

Explanation:

- **None** → Any number of input samples.
- **8** → Every input sentence must have length 8.

---

## Predict on a Single Sentence

```python
model.predict(embedded_docs[0])
```

This tries to predict using **only one sentence**.

However, Keras expects input in the form:

```text
(Number of Samples, Sentence Length)
```

Example:

```text
(1, 8)
```

Instead of

```text
(8,)
```

Therefore, the correct way is:

```python
model.predict(np.array([embedded_docs[0]]))
```

or

```python
model.predict(embedded_docs[0].reshape(1, 8))
```

This converts one sentence into a batch containing one sample.

---

## Workflow

```text
Sentence
      ↓
One-Hot Encoding
      ↓
Integer IDs
      ↓
Padding
      ↓
Embedding Layer
      ↓
Dense Word Vectors
      ↓
RNN / LSTM / GRU
```

---

## Key Points

- The **Embedding Layer** converts integer word IDs into dense vectors.
- `voc_size` defines the number of unique words.
- `dim` defines the size of each word vector.
- `input_length` specifies the sentence length.
- Input shape = `(Number of Sentences, Sentence Length)`.
- Output shape = `(Number of Sentences, Sentence Length, Embedding Dimension)`.
- The Embedding Layer learns these vectors automatically during training.
