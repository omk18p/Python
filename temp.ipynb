```python
def cal_tot(*args,**kwargs):
    print(f'Prices: {args}')
    print(f'Total : {sum(args)}')
    print(f'Customer Details: {kwargs}')

cal_tot(100, 200, 150,
    name="Omkar",
    city="Sangli")
```

```python
def analyze_numbers(*args,**kwargs):
    print(f'numbers : {args}')
    print(f'Even Numbers: {[num for num in args if num%2==0]}')
    print(f'Odd Numbers: {[num for num in args if num%2==1]}')
    print(f'Maxium : {max(args)}')
    print(f'Minimum : {min(args)}')
    print(f'Sum : {sum(args)}')

analyze_numbers(10, 15, 22, 7, 30, 9)


```

```python
def create_profile(*args,**kwargs):
    print(f'Name : {kwargs["name"]}')
    print(f'City : {kwargs["city"]}')
    print(f'Skills : {args}')
    print(f'Total Skills : {len(args)}')
    print(f'Profile Details : {kwargs}')
    
create_profile(
    "Python", "SQL", "Excel", "NumPy",
    name="Omkar",
    age=22,
    city="Sangli"
)


```
