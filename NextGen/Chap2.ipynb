1.  Write an if-elif-else program that takes a number and classifies it as 'Positive', 'Negative', or 'Zero'

```python
no=int(input("Enter Number"))

if no==0:
    print("Zero")
elif no>0:
    print("Positive")
else:
    print("Negative")        
```

4.  Use enumerate() to print every character of a given string alongside its 1-based position (e.g., '1: H', 
'2: e', ...).

```python
text = "Omkar"

for index, char in enumerate(text, start=1):
    print(f"{index}: {char}")
```

### enumerate()

The `enumerate()` function returns both the index and value of each element while iterating over a sequence.
By default, the index starts at 0, but we can use `start=1` to begin from 1.
It is useful when we need both the position and the element in a loop.

5.  Given two lists — subjects and marks — use zip() to iterate over them together and print each pair as 
'Subject: Marks'. 

```python
subjects=["Maths","Sci","IT","OS"]
marks=[80,56,65,89]

for sub,mark in zip(subjects,marks):
    print(sub,":",mark)
```

6.  Given a list of integers, write a loop that uses continue to skip printing negative numbers, and break 
to stop the loop entirely as soon as a number greater than 100 is found.

```python
nos=[10,-11,32,-32,342,34,75,5]

for num in nos:
    if num>=100:
        break
    elif num<0:
        continue
    else:
        print(num)    
```

7.  Write a program using nested for loops to print a right-angled triangle of stars (*), where the number 
of rows is provided by the user. 

```python
rows = int(input("Enter number of rows: "))

for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()
```

8.  Implement the classic FizzBuzz problem for numbers 1 to 50: print 'Fizz' for multiples of 3, 'Buzz' for 
multiples of 5, 'FizzBuzz' for multiples of both, and the number itself otherwise. 

```python
for i in range(1,51):
    if i%3==0 and i%5==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")   
    else:
        print(i)    
        
```

9.  Write a program that searches for a target value inside a list using a for-else construct: if the value is 
found, print its index and break; if the loop completes without finding it, the else clause should print 
'Not found'.

```python
numbers = [10, 20, 30, 40, 50]
target = int(input("Enter target: "))

for inx,num in enumerate(numbers):
    if num==target:
        print(f"Present at index : {inx}")
        break
else:
        print("Not found")    
```

10.  Given a list of integers and a target sum, write a program using nested loops that finds all unique 
pairs of numbers whose sum equals the target — making sure (2,3) and (3,2) are not both counted

```python
numbers = [1, 2, 3, 4, 5, 6]
target = int(input("Enter target sum: "))

for i in range(len(numbers)):
    for j in range(i+1,len(numbers)):
         if numbers[i] + numbers[j] == target:
            print(numbers[i], numbers[j])

```
