8.  Write a program that takes a temperature value and a unit ('C' or 'F') and uses a ternary (conditional) 
expression to convert it to the other unit. Format the result to 2 decimal places.

```python
temp = float(input("Enter temperature: "))
unit = input("Enter unit (C/F): ").upper()

result = (temp * 9/5 + 32) if unit == "C" else ((temp - 32) * 5/9)

print(f"Converted Temp : {result:.2f}")
```

10.  Using only bitwise operators (no % allowed), write an expression that determines whether a given 
integer n is even or odd. Explain in a comment why the bitwise trick works in terms of binary 
representation. 


```python
n = int(input("Enter an integer: "))

if n & 1:
    print("Odd")
else:
    print("Even")
```
